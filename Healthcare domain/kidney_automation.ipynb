```python
import numpy as np
import pandas as pd

data = pd.read_csv('/content/cleaned_kidney_data.csv')
data.head()
```

```python
# taking training & testing dataset
training = data.sample(frac = 0.8, random_state=42).reset_index(drop=True)
testing = data.drop(training.index).reset_index(drop=True)
```

```python
from pycaret.classification import *
```

```python
# model with training data
exp_classification = setup(data = training, target = 'classification', session_id=123)
# if need to change computation mtd of any below features add that in the above code eg. numeric_imputation = 'median'
```

```python
# comparing models
compare_models()
```

```python
#(et) extra tree classifier model shows high score in accuracy & other parameters. so building a model with that
et = create_model('et')
```

```python
# to fine-tune the model further
tuned_et = tune_model(et)
```

```python
# Evaluate the original model
evaluate_model(et)
```

```python
# predict the test dataset
test_prediction = predict_model(et, data=testing)
test_prediction.head()
```

we use .py (in app.py)as its python code. jupyternotebook deosnt support .py file and streamlit doesnt support .ipynb format

✔️ why use @st.cache()

When you use Streamlit to build apps, every time you interact with the app (like uploading a file, clicking a button), Streamlit reruns the script top to bottom to reflect changes.

If you have expensive operations like:

Loading large datasets

Training models

Downloading data from the web

Running heavy computations

These will re-run every time, which is slow.To avoid rerunning costly operations, we use: @st.cache()

✔️ Now, Streamlit remembers (caches) the result, and reuses it unless the function inputs change.


```python
%%writefile app.py
import pickle

import streamlit as st  # frontend code
# loading the trained model
pickle_in = open('kidney_et_model.pkl', 'rb')
et = pickle.load(pickle_in)

# no need for this code here as data is already encoded. but if there was any categorical variables not encoded can use this code

@st.cache()

def prediction('rbc_normal', 'pc_normal', 'pcc_present', 'ba_present', 'htn_yes',
       'dm_yes', 'cad_yes', 'appet_poor', 'pe_yes', 'ane_yes', 'age', 'bp',
       'sg', 'al', 'su', 'bgr', 'bu', 'sc', 'sod', 'pot', 'hemo', 'pcv', 'wc',
       'rc'):
  # making prediction
  pred = et.predict([['rbc_normal', 'pc_normal', 'pcc_present', 'ba_present', 'htn_yes',
       'dm_yes', 'cad_yes', 'appet_poor', 'pe_yes', 'ane_yes', 'age', 'bp',
       'sg', 'al', 'su', 'bgr', 'bu', 'sc', 'sod', 'pot', 'hemo', 'pcv', 'wc',
       'rc']])
  if prediction == 0:
    pred = "No  kidney anomaly found"
  else:
    pred = "Likely Kidney anomaly detected"

 return pred


# Write streamlit function now
def main():
    html_temp= """
    <div style ="background-color:cyan:padding:13px">
    <h1 style="color:black;text-align:center;">Kidney Disease Prediction</h1>
    </div>
    """
    st.markdown(html_temp, unsafe_allow_html=True)
    htn = st.selectbox('htn',("no","yes"))
    dm= st.selectbox('dm status',('no','yes'))
    rbc = st.selectbox('rbc',('normal','abnormal'))
    pc = st.selectbox('pc',('normal','abnormal'))
    pcc = st.selectbox('pcc',('present','notpresent'))
    ba = st.selectbox('ba',('present','notpresent'))
    cad = st.selectbox('cad',('yes','no'))
    appet = st.selectbox('appet',('good','poor'))
    pe = st.selectbox('pe',('yes','no'))
    ane = st.selectbox('ane',('yes','no'))
    age=st.number_input("Age")
    bp = st.number_input("BP")
    sg = st.number_input("SG")
    al = st.number_input("AL")
    su = st.number_input("SU")
    bu = st.number_input("BU")
    sc = st.number_input("SC")
    sod = st.number_input("SOD")
    pot = st.number_input("POT")
    hemo = st.number_input("HEMOGLOBIN")
    bgr = st.number_input("BGR")
    wc=st.number_input("WC")
    pcv = st.number_input("PCV")
    rc = st.number_input("RC")
    RESULT=""


    # WHEN 'PREDICT' is clicked, make the prediction and store it
    if st.button("Predict"):
        result = prediction(et)
        st.success("Report Results: {}".format(result))

if __name__=='__main__':
    main()

```
